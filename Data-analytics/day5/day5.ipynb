### SETUP IMPORTS & PLOTTING CONFIGURATION

```python
import numpy as np
import matplotlib.pyplot as plt
print("hello")

#Apply clean visualization style
plt.style.use('seaborn-v0_8_whitegrid' if 'seaborn-v0_8_whitegrid' in plt.style.available else 'default')
#seaborn-v0_8_whitegrid : a clean white grid style inspired by seaborn's whitegrid theme.
plt.rcParams["figure.dpi"] = 100
# rc = runtime configuration
```

## 1 Linear Spacing


```python
## retstep = true to return the step size along with with generated array(sample , step_size)
x_linear,step_size = np.linspace(start=0.0, stop=10.0,num=15,retstep=True)
print("Linear space array :", x_linear)
#plot the linear space array
plt.figure(figsize=(9,2.5))

# scatter plot parameters:
plt.scatter(x_linear,np.zeros_like(x_linear),color='red',s=60,zorder=3,label=f'Linear Space (h={step_size:.2f})')

plt.vlines(x_linear, ymin=-0.1,ymax=0.1,colors='black',alpha=0.5,linestyles='--')

plt.title("Linear Space with vertical Line ", fontsize=12,fontweight='bold')
plt.xlabel("Value")
plt.xticks(np.arange(0, 11, 2))
plt.yticks([])
plt.ylim(-0.5,0.5)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

## Logarithmic Spacing(np.logspace)

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 7 points from 10^-5 to 10^1 using logarithmic spacing
learning_rates = np.logspace(
    start=-5,
    stop=1,
    num=7,
    base=10
)

print("Logarithmic space array:", learning_rates)

print("Generated Hyperparameters: learning_rates")

print("Learning rates:", [f"{lr:.6f}" for lr in learning_rates])


# Create 2 plots
fig, (ax1, ax2) = plt.subplots(
    nrows=2,
    ncols=1,
    figsize=(8, 4.5)
)


# ---------------- TOP PLOT ----------------
# Logarithmic scale visualization

ax1.scatter(
    learning_rates,
    np.zeros_like(learning_rates),
    color='pink',
    s=60,
    zorder=3,
    label='Learning Rates'
)

ax1.set_xscale('log')

ax1.set_title(
    "Logarithmic Space of Learning Rates",
    fontsize=12,
    fontweight='bold'
)

ax1.set_xlabel("Value (log scale)")
ax1.set_yticks([])
ax1.legend()


# ---------------- BOTTOM PLOT ----------------
# Linear scale visualization

ax2.scatter(
    learning_rates,
    np.zeros_like(learning_rates),
    color='purple',
    s=60,
    zorder=3,
    label='Learning Rates'
)

ax2.set_xscale('linear')

ax2.set_title(
    "Logarithmic Values on Linear Scale",
    fontsize=12,
    fontweight='bold'
)

ax2.set_xlabel("Value (linear scale)")
ax2.set_yticks([])
ax2.legend()


plt.tight_layout()
plt.show()
```
