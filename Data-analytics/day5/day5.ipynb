### SETUP IMPORTS & PLOTTING CONFIGURATION

```python
import numpy as np
import matplotlib.pyplot as plt
print("hello")

#Apply clean visualization style
plt.style.use('seaborn-v0_8_whitegrid' if 'seaborn-v0_8_whitegrid' in plt.style.available else 'default')
#seaborn-v0_8_whitegrid : a clean white grid style inspired by seaborn's whitegrid theme.
plt.rcParams["figure.dpi"] = 100
# rc = runtime configuration
```

## 1 Linear Spacing


```python
## retstep = true to return the step size along with with generated array(sample , step_size)
x_linear,step_size = np.linspace(start=0.0, stop=10.0,num=15,retstep=True)
print("Linear space array :", x_linear)
#plot the linear space array
plt.figure(figsize=(9,2.5))

# scatter plot parameters:
plt.scatter(x_linear,np.zeros_like(x_linear),color='red',s=60,zorder=3,label=f'Linear Space (h={step_size:.2f})')

plt.vlines(x_linear, ymin=-0.1,ymax=0.1,colors='yellow',alpha=0.5,linestyles='--')

plt.title("Linear Space with vertical Line ", fontsize=12,fontweight='bold')
plt.xlabel("Value")
plt.xticks(np.arange(0, 11, 2))
plt.yticks([])
plt.ylim(-0.5,0.5)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```
