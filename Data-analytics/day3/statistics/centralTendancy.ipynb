```python
import numpy as np
import matplotlib.pyplot as plt
incomes = np.random.normal(loc=10.0,scale=20.0,size=1000)
plt.hist(incomes,bins=50)
plt.show()
```

```python
print("mean:",np.mean(incomes),sep="\t:\t")
print("median:",np.median(incomes),sep="\t:\t")
```

```python
ages = np.random.randint(18, 81, size=500)
print("ages : " ,ages,sep="\t:\t")
```
