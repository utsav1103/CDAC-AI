## Logarithmic Spacing(np.logspace)

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 7 points from 10^-5 to 10^1 using logarithmic spacing
learning_rates = np.logspace(
    start=-5,
    stop=1,
    num=7,
    base=10
)

print("Logarithmic space array:", learning_rates)

print("Generated Hyperparameters: learning_rates")

print("Learning rates:", [f"{lr:.6f}" for lr in learning_rates])


# Create 2 plots
fig, (ax1, ax2) = plt.subplots(
    nrows=2,
    ncols=1,
    figsize=(8, 4.5)
)


# ---------------- TOP PLOT ----------------
# Logarithmic scale visualization

ax1.scatter(
    learning_rates,
    np.zeros_like(learning_rates),
    color='pink',
    s=60,
    zorder=3,
    label='Learning Rates'
)

ax1.set_xscale('log')

ax1.set_title(
    "Logarithmic Space of Learning Rates",
    fontsize=12,
    fontweight='bold'
)

ax1.set_xlabel("Value (log scale)")
ax1.set_yticks([])
ax1.legend()


# ---------------- BOTTOM PLOT ----------------
# Linear scale visualization

ax2.scatter(
    learning_rates,
    np.zeros_like(learning_rates),
    color='purple',
    s=60,
    zorder=3,
    label='Learning Rates'
)

ax2.set_xscale('linear')

ax2.set_title(
    "Logarithmic Values on Linear Scale",
    fontsize=12,
    fontweight='bold'
)

ax2.set_xlabel("Value (linear scale)")
ax2.set_yticks([])
ax2.legend()


plt.tight_layout()
plt.show()
```

## 4 STEP BASED RANGE(np.arange)

```python
time_steps=np.arange(start=0.0,stop=5.0,step=0.5)
print("Time Steps Array: ", time_steps)

print("Common difference of time steps array: ", time_steps[1]-time_steps[0])
print("Total number of time steps: ", len(time_steps))


#Graphical Visualization of time steps
plt.figure(figsize=(9,2.5)) #figsize=(width,height) in inches

#Stem plot Details:
# Stem plot: a discrete plot that displays data points as stems extending from a baseline along the x-axis. Each stem represents a data point, and the height of the stem corresponds to the value of that point. It is useful for visualizing discrete data or events over time.
#linefmt: "C4-" specifies the color and style of the stem lines (C4 is a color from the default color cycle, and '-' indicates a solid line).
#color cycle definations: C1-blue, C2-orange, C3-green, C4-red, C5-purple, C6-brown, C7-pink, C8-gray, C9-olive, C10-cyan

#markerfmt: "C4o" specifies the color and style of the markers at the top of each stem (C4 is red, and 'o' indicates circular markers).
#basefmt: "k-" specifies the color and style of the baseline (k is black, and '-' indicates a solid line).


plt.stem(time_steps,np.ones_like(time_steps),linefmt='C4--',markerfmt='C4o',basefmt='')

plt.title("Time Steps Visualization", fontsize=16)
plt.xlabel("Time Steps", fontsize=14)
plt.yticks([]) #remove y-ticks
plt.ylim(0,1.3)
plt.tight_layout() #adjust subplots to fit into figure area.
plt.show() 
```

## Coordinate Grid(np.meshgrid)

```python
x_domain=np.linspace(-2,2,20)
y_domain = np.linspace(-2,2,20)

#Construct 2D coordinate matrix
X,Y = np.meshgrid(x_domain,y_domain)

#Evaluate a 2D paraboloid surface Z=X^2 + Y^2
Z=X**2 + Y**2

print("Grid Coordinate output")
print(f"X Matrix",{X.shape})
print(f"Y Matrix",{Y.shape})
print(f"Z Matrix",{Z.shape})
print("sample X matrix slice(first 3 X 3 block):\n",X[:3,:3])
print("sample Y matrix slice(first 3 X 3 block):\n",Y[:3,:3])
print("sample Z matrix slice(first 3 X 3 block):\n",Z[:3,:3])


#Graphical Visualization of 2D paraboloid surface
fig , (ax1,ax2) = plt.subplots(nrows=1,ncols=2,figsize=(11,4.5))
# Subplot 1
ax1.scatter(X,Y,color='blue',marker='o',s=60,edgecolor='k') #s=marker size
ax1.set_title("2D Coordinate Grid mesh(np.meshgrid)", fontsize=11)
ax1.set_xlabel("X-axis", fontsize=11)
ax1.set_ylabel("Y-axis", fontsize=11)   
ax1.set_aspect('equal', ) #set equal aspect ratio for x and y axes

# Subplot 2
countour = ax2.contourf(X,Y,Z,cmap='viridis',levels=15) #filled contour plot

# ax2.contour(X,Y,Z,colors='k',levels=15,linewidths=0.5) #add contour lines to the filled contour plot

fig.colorbar(countour, ax=ax2, label='Z values = f(X,Y)') #add colorbar to the contour plot
ax2.set_title("2D Contour Plot of Paraboloid Surface", fontsize=11)
ax2.set_xlabel("X-axis", fontsize=11)   
ax2.set_ylabel("Y-axis", fontsize=11)
ax2.set_aspect('equal', ) #set equal aspect ratio for x and y axes

plt.tight_layout() #adjust subplots to fit into figure area.
plt.show()
```

## Trigonometric

```python
t = np.linspace(
    start = 0,
    stop = 10,
    num = 200
)

f = 2.0
A = 3.0

y_pure = A * np.sin( 2 * np.pi * f * t)

gaussian_noise = np.random.normal(
    loc=0.0,
    scale=0.6,
    size=len(t)
)

y_noisy = y_pure + gaussian_noise

print(" Harmonic signal generated successfully with gaussian noise.")
print("Time Domain Array length:",len(t))
print(f"Pure signal range :[{y_pure.min():.2f},{y_pure.max():.2f}]")
print(f"Noisy signal range :[{y_noisy.min():.2f},{y_noisy.max():.2f}]")
print(f"Calculated Noise variance : {np.var(gaussian_noise):.4f}")

plt.figure(figsize=(10, 4.5))

# noisy observation (scatter plot)
plt.scatter(
    x=t,
    y=y_noisy,
    color='red',
    s=14,
    alpha=0.55,
    edgecolors='none',
    zorder=2,
    label=r'Noisy Signal Observation ($\sigma=0.6$)'
)

# pure signal
plt.plot(
    t,
    y_pure,
    color='grey',
    linewidth=2.8,
    zorder=3,
    label=r'Pure Signal $y = 3\sin(4\pi t)$'
)

plt.title(
    label="Harmonic Wave with Additive Gaussian Noise",
    fontsize=13,
    fontweight='bold'
)

plt.xlabel(
    xlabel='Time (seconds)',
    fontsize=11
)

plt.ylabel(
    ylabel='Amplitude',
    fontsize=11
)

plt.grid(
    True,
    linestyle='--',
    alpha=0.25
)

plt.legend(
    loc='upper right',
    frameon=True,
    fontsize=9
)

plt.tight_layout()
plt.show()
```

## 2. Exponential Decay + Uniform Measurement Error

# overview & purpose

Simulates physical decay dynamics (e.g radioactive decay capacitor discharge thermal cooling) subjected to bounded quantization or sensor tolerance error modeled by a uniform distribution.


```python
# generate decay signal

t_decay = np.linspace(
    start=0.0,
    stop=5.0,
    num=100    
)

N0 = 10.0
decay_rate = 0.8

y_decay_pure = N0 * np.exp(-decay_rate * t_decay)

uniform_noise = np.random.uniform(
    low=-0.5,
    high=0.5,
    size = len(t_decay)
)

y_decay_measured = y_decay_pure + uniform_noise

print("=== 2. Exponential decay + uniform error ===")
print(f"Initial pure value at t=0: {y_decay_pure[0]:.2f}")
print(f"final pure value at t=5: {y_decay_pure[-1]:.4f}")
print(f"uniform error bounds: [{uniform_noise.min():.4f}, {uniform_noise.max():.4f}]")

plt.figure(figsize=(9,3.5))

plt.fill_between(
    x=t_decay,
    y1=y_decay_pure - 0.5,
    y2=y_decay_pure + 0.5,
    color='lightgray',
    alpha=0.5,
    label='Uniform noise bounds($\\pm 0.5$)' 
)

plt.plot(
    t_decay,
    y_decay_measured,
    's-',
    color='green',
    linewidth=1,
    markersize=4,
    alpha=0.8,
    label='Measured data'
)

plt.plot(
    t_decay,
    y_decay_pure,
    color='blue',
    linewidth=2.5,
    label='Theoretical decay $10e^{-0.8t}$'
)

plt.title(
    label='Exponential decay curve with bounded uniform error',
    fontsize=12,
    fontweight='bold'
)

plt.xlabel(xlabel='Time(t)')
plt.ylabel(ylabel='Quantity / signal level')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

# polynomial function + sparse outliers (salt & pepper / impulse Noise)


## overview & purpose


```python
x_poly = np.linspace(
    start=-2.0,
    stop=2.0,
    num=120
)

#cubic polynomial : p(x) = 0.5*x^3 - x^2 - x + 2 

y_poly_clean = 0.5 * (x_poly**3) - (x_poly**2) - x_poly + 2
y_poly_corrupted = y_poly_clean.copy()

#select 8% of indices randomly to inject impulsive spikes

num_outliers = int(0.08 * len(x_poly))

outlier_indices = np.random.choice(a=len(x_poly),
    size=num_outliers,
    replace=False               
)

spikes_magnitude = np.random.uniform(
    low=4.0,
    high=8.0,
    size=num_outliers
)

y_poly_corrupted[outlier_indices] += spikes_magnitude

print("=== 3. PolyNomial + ImpuLsE Outliers ===")
print(f"total data points: {len(x_poly)}")
print(f"Injected outlier count: {num_outliers}")
print(f"Outlier indices: {outlier_indices}")
print(f"Outlier values: {y_poly_corrupted[outlier_indices]}")
print(f"Spike magnitudes: {spikes_magnitude}")
print(f"Clean data mean: {np.mean(y_poly_clean):.4f}")


plt.figure(figsize=(9,3.5))
plt.plot(
    x_poly,
    y_poly_clean,
    label="Clean Polynomial",
    color='black',
    linewidth=2,
    zorder=2
)

normal_mask = np.ones_like(x_poly,dtype=bool)
normal_mask[outlier_indices] = False 

plt.scatter(
    x = x_poly[normal_mask],
    y = y_poly_corrupted[normal_mask],
    label='normal points',
    color='green',
    s=20,
    zorder=4,
    alpha=0.6
)
plt.scatter(
    x = x_poly[outlier_indices],
    y = y_poly_corrupted[outlier_indices],
    label='Impulse outliers',
    color='blue',
    s=20,
    zorder=4,
    alpha=0.6,
    marker='X'
)

plt.title(
    label='Polynomial with impulse outliers',
    fontsize=12,
    fontweight='bold'
)
plt.xlabel(xlabel="x Domain")
plt.ylabel(ylabel="y range")
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```
